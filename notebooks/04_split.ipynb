```python
import pandas as pd

df = pd.read_csv("../data/processed/landmarks.csv")
print("Shape:", df.shape)
print("Unique signers:", df["signer_id"].nunique())
df["signer_id"].value_counts().head()
```

```python
import numpy as np

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

all_signers = df["signer_id"].unique()
np.random.shuffle(all_signers)

n_test_signers = int(len(all_signers) * 0.2)
test_signers = set(all_signers[:n_test_signers])
train_signers = set(all_signers[n_test_signers:])

print("Total signers:", len(all_signers))
print("Train signers:", len(train_signers))
print("Test signers:", len(test_signers))
```

```python
train_df = df[df["signer_id"].isin(train_signers)].reset_index(drop=True)
test_df = df[df["signer_id"].isin(test_signers)].reset_index(drop=True)

print("Train rows:", len(train_df))
print("Test rows:", len(test_df))
print("Test set %:", round(len(test_df) / len(df) * 100, 1))

train_df.to_csv("../data/processed/train.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)
print("\nSaved train.csv and test.csv")
```

```python
train_classes = set(train_df["label"].unique())
test_classes = set(test_df["label"].unique())
all_classes = set(df["label"].unique())

missing_from_train = all_classes - train_classes
missing_from_test = all_classes - test_classes

print("Total classes:", len(all_classes))
print("Train me maujood classes:", len(train_classes))
print("Test me maujood classes:", len(test_classes))
print("\nTrain se ghayab:", missing_from_train if missing_from_train else "Koi nahi ✓")
print("Test se ghayab:", missing_from_test if missing_from_test else "Koi nahi ✓")

# har class ki train aur test me kitni images hain, khaas taur par Tay dekho
print("\nTay ka breakdown:")
print("  Train:", (train_df["label"] == "Tay").sum())
print("  Test:", (test_df["label"] == "Tay").sum())
```

```python
train_files = set(train_df["filename"] + "_" + train_df["label"])
test_files_set = set(test_df["filename"] + "_" + test_df["label"])

overlap = train_files & test_files_set

print("Train files:", len(train_files))
print("Test files:", len(test_files_set))
print("Overlap (dono me maujood):", len(overlap))
print("Test pass?", len(overlap) == 0)

# signer overlap bhi ek dafa pakka check kar lete hain
signer_overlap = train_signers & test_signers
print("\nSigner overlap:", len(signer_overlap))
```
