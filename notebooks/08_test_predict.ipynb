```python
import sys
sys.path.append("..")

from src.predict import predict
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../data/raw/UAlpha40 A Comprehensive Dataset of Urdu alphabets for Pakistan Sign Language")
test_df = pd.read_csv("../data/processed/test.csv")

# 10 random test images test karte hain
sample = test_df.sample(10, random_state=42)

correct = 0
for _, row in sample.iterrows():
    img_path = RAW_DIR / f"{row['label']}-Original" / row["filename"]
    result = predict(img_path)

    is_correct = result["letter"] == row["label"]
    correct += is_correct
    status = "✓" if is_correct else "✗"

    print(f"{status} Asal: {row['label']:10s} | Predict: {result['letter']:10s} | Confidence: {result['confidence']:.2%}")

print(f"\n{correct}/10 sahi")
```
