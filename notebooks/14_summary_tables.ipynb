```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
R = ROOT / "results"

# --- on test.csv (report only, the choice was made by CV) ---
test_a = pd.read_csv(R / "advanced_features_test.csv")      # core v1, v2, v3, v3b
test_b = pd.read_csv(R / "advanced_v4_test.csv")            # v3, v4
test_all = pd.concat([test_a, test_b[test_b.model.str.startswith("v4")]], ignore_index=True)
test_tbl = test_all.pivot_table(index="model", columns="hand", values="test_acc", sort=False).round(4)
print("TEST accuracy (right / left hand)")
print(test_tbl, "\n")

# --- signer-wise CV (on train.csv, the choice was made from this) ---
cv_a = pd.read_csv(R / "advanced_grouped_cv.csv")           # A, B, C, D
cv_b = pd.read_csv(R / "advanced_v4_cv.csv")                # C, F
cv_all = pd.concat([cv_a, cv_b[cv_b.variant.str.startswith("F")]], ignore_index=True)
cv_tbl = cv_all.pivot_table(index="variant", columns="hand", values="acc", sort=False).round(4)
print("Signer-wise CV accuracy (right / left hand)")
print(cv_tbl)

test_tbl.to_csv(R / "summary_test_variants.csv")
cv_tbl.to_csv(R / "summary_signer_cv_variants.csv")
```
